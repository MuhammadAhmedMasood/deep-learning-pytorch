### matrix multiplication:

two main ways to perform multiplication in NN and deep learning:

1. Element-wise multiplication
2. Matrix multiplication (dot product e.g a.b)

There are two rules that matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:
* `(3,2) @ (3,2)` won't work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work
2. The resulting matrix has the shape of the **outer dimensions**:
* `(2,3) @ (3,2) -> (2,2)`
* `(3,2) @ (2,3) -> (3,3)`

```python
# element-wise multiplication:
print(f"{tensor}*{tensor}")
print(tensor*tensor)
```

```python
#matrix multiplication - this does a dot product
torch.matmul(tensor, tensor)
```

### one of the most common errors in DL is shape errors

```python
# shapes for matrix multiplication:
tensor_A = torch.tensor([[1,2],
                         [3,4],
                         [5,6]])
tensor_B = torch.tensor([[7,8],
                         [9,10],
                         [11,12]])
```

```python
tensor_A.shape
```

```python
tensor_B.shape
```

```python
# torch.mm(tensor_A, tensor_B) # torch.mm is same as torch.matmul (it's an alias for matmul)
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a transpose:

**transpose** switches the axes or dimensions of a given tensor

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# matrix multiplication works when tensor_B is transposed

print(f"Original shapes: tensor_A = {tensor_A.shape} and tensor_B = {tensor_B.shape}")
print(f"New Shapes: tensor_A = {tensor_A.shape} (same shape as above) and tensor_B.T = {tensor_B.T}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match")
print("Output: \n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\nOutput Shape: ", output.shape)
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
# let's create a tensor
y = torch.arange(1., 10.)
y, y.shape
```

```python
y.dtype
```

```python
# add extra dimension
# y_reshaped = y.reshape(1, 7) # this will give error as we are trying to squeeze 9 elements into 9
y_reshaped = y.reshape(1, 9) # 1 row and 9 cols
```

```python
# change the view
z = y.view(1,9) # view similar to reshape, but shares the same memory of the tensor so z and y are pointing to the same memory
z, z.shape
```

```python
# changing z changes y because a view of a tensor shares the same memory as the original
z[:, 0] = 5
z, y, y_reshaped
```

```python
# stack tensor on top of each other
y_stacked = torch.stack([y,y,y,y], dim = 0) # takes a list of tensors - dim = 0 is vstack and dim=1 is hstack
y_stacked
```

```python
y_ver_stack = torch.vstack([y,y])
y_ver_stack
```

```python
y_hor_stack = torch.hstack([y,y])
y_hor_stack
```

```python
y_reshaped, y_reshaped.shape
```

```python
# squeeze (removes all single dimensions from the tensor) and unsqueeze:
y_squeeze = torch.squeeze(y_reshaped)
y_squeeze
```

```python
y_squeeze.shape # 1 dimension removed
```

```python
# unsqueeze - adds a single dimension to a target tensor
y_unsqueeze = torch.unsqueeze(y_squeeze, dim = 0) # can give dim = 1
y_unsqueeze, y_unsqueeze.shape
```

```python
y_unsqueeze_2 = torch.unsqueeze(y_unsqueeze, dim = 2) # can give dim = 1
y_unsqueeze_2, y_unsqueeze_2.shape
```

```python
# torch.permute - rearranges the dimensions in a specified order - returns a view
x_original = torch.rand(size = (224,224,3)) # [heights, width, colour_channels]

# permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # takes in dims as input - shifts axis 0 -> 1, 1 -> 2, 2 -> 0

print(f"Prevous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

```python
# one thing is like view, x_permuted and x_original are pointing to the same memory so let's check:
x_original[0,0,0] = 694848
x_original
```

```python
x_permuted[0,0,0] # so changed
```

```python
x_original.dtype
```

```python
x_permuted.dtype
```

## Reproducibility: trying to take random out of random

In short how a neural network learns:

`start with random number -> tensor operations -> update random numbers to try and make better representations of the data -> again -> again -> again...`

To reduce the randomness in NN and Pytorch comes the concept of a **random seed**

Essentially what the random seed does is "flavour" the randomness.

```python
# create two random tensors
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```
