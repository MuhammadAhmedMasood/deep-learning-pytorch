### Random Tensors:

Random tensors are impt because the way many NN learn is that they starts with tensors full of random numbers and then adjust those random numbers to better represent the data.

`Start with random number -> look at data -> update random numbers -> look at data -> update random numbers`

```python
# create a random tensors of size (3,4)

random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size = [3, 224,224]) # height, width, colour channels (R, G, B)
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

```python
r_tensor = torch.rand(3,3,4,4,)
r_tensor
```

```python
r_tensor.shape
```

```python
r_tensor.ndim
```

### Creating a range of tensors and tensors-like:

```python
# use torch.arange():

one_to_ten = torch.arange(1,11,1)
one_to_ten
```

```python
# creating tensors-like:
ten_zeros = torch.zeros_like(input = one_to_ten)
ten_zeros
```

```python
one_to_ten.dtype
```

### tensors datatypes:

**Notes:** Tensor datatypes one of the 3 big erros you'll run unto with pytorch and deep learning:
1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device

```python
# Float 32 tensor:
float_32_tensor = torch.tensor([3.0,6.0,9.0],
                               dtype = None, # what daat type of the tensor: float32 or float 16 etc
                               device = None, # by default it is cpu, can change to cuda - tensors must be on same device when we perform operations
                               requires_grad = False) # if we want pytorch to track gradients when tensors go into certain calculations
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensors = float_32_tensor.type(torch.float16)
float_16_tensors
```
